```python
df1 = pd.DataFrame({
    'key': [1, 2, 3],
    'Name': ['Anu', 'Rahul', 'Maya']
})

df2 = pd.DataFrame({
    'key': [1, 2, 3],
    'Marks': [85, 90, 88]
})
```

```python
pd.merge(df1, df2, on='key')
```

```python
df1.join(df2,lsuffix='_left',rsuffix='_right')
```
